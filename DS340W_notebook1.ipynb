```python
import pandas as pd
from sklearn.model_selection import train_test_split
import numpy as np
```

```python
# Read in mlb_dv dataset
mlb_dv = pd.read_csv('mlb_dv.csv')
mlb_dv.head()
```

```python
# Split into train, test, validation - 70, 20, 10
n = len(mlb_dv)
train_end = int(n * 0.70)
val_end = int(n * 0.80)  # 70% + 10%

train = mlb_dv.iloc[:train_end]
val = mlb_dv.iloc[train_end:val_end]
test = mlb_dv.iloc[val_end:]
```

```python
# Show number of rows of each dataset
print(len(train), 'train examples')
print(len(test), 'test examples')
print(len(val), 'validation examples')
```

```python
# Export each dataset to csv
pd.DataFrame(train).to_csv('mlb_dv_train.csv')
pd.DataFrame(test).to_csv('mlb_dv_test.csv')
pd.DataFrame(val).to_csv('mlb_dv_val.csv')
```
